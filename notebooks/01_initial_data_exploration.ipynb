## Data Diagnosis

### Matrix Dimension
We are working with 13,711 observations x 8 variables.

### Data Types
Data types are correctly assigned for almost every variable:
- `Año` and `Trimestre` are `int64`, which is ideal for future exploratory analysis.
- `Valor` is `float64`, allowing us to apply descriptive statistics.
- `Régimen`, `Provincia`, and `Comunidad_Autónoma` are `str`.
- `CPRO` and `CODAUTO` are `float64` because data is missing in some rows, forcing pandas to convert the whole column to `float64` (Python integers cannot hold null values natively).

### Missing Data
- `Valor` (844 nulls - 6.1%): There are 844 quarters/zones where the administration did not record prices (or they were not published).
- `CPRO` (3,535 nulls - 25.7%): About 25% of rows do not have a provincial code. This may not be an error; in public datasets, official sources often include summary rows (e.g., 'Total España' or 'Total Andalucía') that do not have a unique province assigned.
- `Comunidad_Autónoma` (1 null): This is the only row in the dataset missing an assigned community. We should investigate the reason behind this.
